# **Paws & Ratings**

### ***Exploratory Data Analysis***

The impact of semantic markers and zoomorphic imagery in media content metadata on users rating and audience engagement (using MyAnimeList database as a case of study).

**Goals**:

1. **Impact Analysis**: Investigate how mentions of cats and dogs in anime titles and synopses affect users ratings and audience engagement.
2. **Comparison**: Compare "cat" and "dog" titles to dotermine which receives higher average scores.
